```python
set1={1,2,3,3,'hello'}
set1.add(5)
set1.add(100)
set1.add('hello')

# delete the element from the set
# 1. remove method delete the element from the set and If you are trying to delete an element from the set
# which is not in the element set then remove will give an error
set1.remove('hello')
# set1.remove('hii') # if you are trying to delete 'hii' from the set it will give you error becacuse 'hii' is not persent in the set

# 2 . discard() method -> is also use for delete the element from  the set but an element is not persent 
# in the set but if you are trying to delete that those element then discard not give you an error
set1.discard('100')
set1.discard('hii') # this element is not persent in the set but not give you an error
print(set1)
```

```python
# check if an element is in the set or not
set1={1,2,3,3,'hello'}
if 'hello' in set1:
    print(True)
else:
    print(False)
```

```python
# if you want to the clear the set use clear method
set1.clear()
print(set1)
```

```python
# Copy method in Set 
# 1. Deep Copy
set1 = {1,2,3,4}
set2 = set1
set2.add(4)
print(set1)
print(set2)
```

```python
# operation on set
set1={1,8,5,2,4}
set2={8,2,9,6}

# 1. Union 
print(set1 | set2)
print(set1.union(set2))

# 2. intersection
print(set1 & set2)
print(set1.intersection(set2))

# 3 difference
print(set1 - set2)
print(set1.difference(set2))

print(set2 - set1)
print(set2.difference(set1))

# 4 symmetric difference
print(set1^set2)
print(set1.symmetric_difference(set2))
```
