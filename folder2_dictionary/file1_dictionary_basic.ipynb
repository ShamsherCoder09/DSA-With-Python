```python
dic = {1:'abhinav','hello':100,10.12:'hiii','key':True}
print(dic)
print(type(dic))
print(len(dic))
```

```python
dic = {(1,2,3):123,1:'abhinav','hello':100,10.12:'hiii','key':True}
print(dic)
```

```python
# access the element of the dictionary by using the key
print(dic[1])
print(dic['hello'])
```

```python
# operation on dictionary 
# 1 operation add and update on dictionary
dic1 = {1:'Sam',2:'Sahil',3:'Uvaiz',4:'Yash'}
dic1[5]="Shamsher"
dic1[2] = "kamil"

dic1.update({5:"sher",3:'Rahil'})

print(dic1)
```

```python
# delete the dicionary element by using delete and pop method
dic1 = {1:'Sam',2:'Sahil',3:'Uvaiz',4:'Yash'}
dic1[5]="Shamsher"
dic1[2] = "kamil"

dic1.pop(2)
del dic1[1]

# access the dictionary element by using get()
print(dic1[3])
print(dic1.get(3))
print(dic1)
```

```python
# copy operation in set
# it is deep copy
dic1 = {1:'Sam',2:'Sahil',3:'Uvaiz',4:'Yash'}
dic2 = dic1

dic2[5] = 'Shamsher'
print(dic1)
print(dic2)
```

```python
# if you want to only copy the element called shallow copy
dic1 = {1:'Sam',2:'Sahil',3:'Uvaiz',4:'Yash'}
dic2 = dic1.copy()

dic2[5] = 'Sham'
print(dic1)
print(dic2)
```

```python
# iteration on dictionary
dic1 = {1:'Sam',2:'Sahil',3:'Uvaiz',4:'Yash'}
dic1.update({5:'Shamsher'})

# print key and value
for key, value in dic1.items():
    print(key, value)

# print only key

for key in dic1.keys():
    print(key)

#print only value
for value in dic1.values():
    print(value)
```

```python
# count the frequency by the dictionary
list1 = ['abhinav','aman','abhinav','yash','aman','rahul','abhinav','yash']

freq = {}
for name in list1:
    if name not in freq:
        freq[name] = 1
    else:
        freq[name] +=1
print(freq)
```
